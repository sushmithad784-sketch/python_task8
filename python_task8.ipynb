```python
df["Current_Ratio"]=df["Current_Assets"]/df["Current_Liabilities"]
```

```python
df["ROE"]=(df["Net_Income"]/df["Equity"])*100
```

```python
df["EPS"]=df["Net_Income"]/df["Shares_Outstanding"]
```

```python
df
```

```python
print("/nFinancial Ratio Analysis:")
```

```python
df[["Year","Net_Income","Current_Ratio","ROE","EPS"]]
```

```python
# Max Current Ratio
Max_Ratio = df["Current_Ratio"].max()
print("Max_Ratio:", Max_Ratio)

# Max ROE
Max_ROE = df["ROE"].max()
print("Max_ROE:", Max_ROE)

# Max EPS
Max_EPS = df["EPS"].max()
print("Max_EPS:", Max_EPS)
```

```python
# Min Current Ratio
Min_Ratio = df["Current_Ratio"].min()
print("Min_Ratio:", Min_Ratio)

# Min ROE
Min_ROE = df["ROE"].min()
print("Min_ROE:", Min_ROE)
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Year"], df["Current_Ratio"], marker="o")
plt.title("Current Ratio Trend")
plt.xlabel("Year")
plt.ylabel("Current_Ratio")
plt.grid(True)
plt.show()
```
